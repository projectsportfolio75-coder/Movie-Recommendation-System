```python
def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name']) 
    return L 
```

```python
# In cast column we have to just pick top three actors 
```

```python
def convert3(text):
    L = []
    counter = 0
    for i in ast.literal_eval(text):
        if counter < 3:
            L.append(i['name'])
        counter+=1
    return L 
```

```python
# Cast Column
movies['cast'] = movies['cast'].apply(convert3)
movies.head()
```

```python
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L 
```

```python
def fetch_director(text):
    L = []
    try:
        for i in ast.literal_eval(text):
            if i['job'] == 'Director':
                L.append(i['name'])
    except (ValueError, SyntaxError) as e:
        # If parsing fails, return empty list
        print(f"Error parsing: {text[:100]}...")  # Optional: for debugging
        return []
    return L
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
# Overview column is in string format so converting it into list
type(movies['overview'][0])
```

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split())
movies.head()
```

#  Important Step

Now we have to remove the space btw the texts for eg "Vansh Kumar" to 'VanshKumar' because  as we will do text analysis so machine might get confused btw "Vansh Sharma"  and "Vansh Kumar ".

```python
# User Defined Fn
def collapse(L):
    L1 = []
    for i in L:
        L1.append(i.replace(" ",""))
    return L1
```

```python
movies['cast'] = movies['cast'].apply(collapse)
movies['crew'] = movies['crew'].apply(collapse)
movies['genres'] = movies['genres'].apply(collapse)
movies['keywords'] = movies['keywords'].apply(collapse)
```

```python
movies.head()
```

```python
# Converting all text column into one column named as 'tags'
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
new = movies.drop(columns=['overview','genres','keywords','cast','crew'])
new.head()
```

```python
# Converting list into string
new['tags'] = new['tags'].apply(lambda x: " ".join(x))
new.head()
```

# Dealing with Repetition of Words

```python
# Here the new problem arises 
# for eg ["loved " ,'loving' ,'love']
# If these type of repetitive texts are present in our data then machine might  get confused 
# Applying steming technique 
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
# User Defined Fn
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
        
    return " ".join(y)    
        
        
        
```

```python
# for eg 
ps.stem("loved")
```

```python
new['tags'] = new['tags'].apply(stem)
```

```python
# cv.get_feature_names()
cv.get_feature_names_out()
```

```python
# We will find the vector which is nearest to our choosen/given vector.
# we are measuring differnce in terms of angle.
# We are working on 5000 Dimensional Data so can't take eucledian distance.
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity
```

```python
# It is a matrix.
# Diagonal element always 1.
similarity.shape
```

```python
similarity[0]
```

```python
# How to get the index of a given movie ? 
new[new['title'] == 'The Lego Movie'].index[0]
```

```python
# Final User Defined function for movie recommendation
def recommend(movie):
    index = new[new['title'] == movie].index[0]
    distances = sorted(list(enumerate(similarity[index])),reverse=True,key = lambda x: x[1])
    for i in distances[1:6]:
        print(new.iloc[i[0]].title)
        
    
```

```python
# Taking Input from user 
recommend('Batman Begins')
```
